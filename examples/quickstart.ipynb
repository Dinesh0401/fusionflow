# FusionFlow Quickstart

This notebook demonstrates the `%%fusionflow` cell magic shipped in v0.4.0. Each cell is a complete `.ff` specification; the magic parses it, executes the experiment via the Pandas backend, and returns the metrics as a pandas Series.

**Install**: `pip install fusionflow[jupyter]`

**Note**: launch Jupyter from the repo root so the relative path `../tests/fixtures/tiny.csv` resolves correctly.

```python
%load_ext fusionflow.integrations.jupyter_magic
```

```python
%%fusionflow
dataset customers v1
    source "../tests/fixtures/tiny.csv"
end

pipeline regression_pipe
    from customers v1
    features [age, income]
    split 0.7
    target spend
end

model linear
    type linear_regression
    params { fit_intercept: true }
end

experiment regression_baseline
    uses pipeline regression_pipe
    uses model linear
    metrics [rmse, mae]
end
```

## What just happened?

- The cell content was parsed as `.ff` source
- The first experiment (`regression_baseline`) was selected automatically
- The Pandas backend loaded `tiny.csv`, applied the pipeline, trained `LinearRegression`, and computed `rmse` and `mae`
- Metrics returned as a pandas Series for inline display

Try changing `split 0.7` to `split 0.8` and re-running.
